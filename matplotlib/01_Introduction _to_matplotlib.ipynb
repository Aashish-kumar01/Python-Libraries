```python
import matplotlib.pyplot as plt
```

## Plotting the line using x and y coordinate


```python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y)
plt.show()
```

### set the title and label for the x and y axis


```python
x = ["Mon", "Tue", "Wed", "Thru", "Fri"] # X-axis
y = [10, 15, 7, 20, 12] # Y-axis

plt.plot(x,y)
plt.title("Bakery Sales this week")

plt.xlabel("Day of the week")
plt.ylabel("Sales per day")

plt.show()
```

### Exploring the plt methods like (legend, xlim, ylim, xticks, etc)

```python
#plt.plot(x, y, color="", linestyle ="linestyle", linewidth=value, marker="marker symbol", label="label name")
#plt.xlabel("text")
#plt.xlim(start, end)


months = [1, 2, 3, 4]
sales = [1000, 1500, 1200, 1002]

plt.plot(months, sales, color= "red", linestyle="--", linewidth= 2, marker ='o', label = "2025 sales data")
plt.xlabel("Months")
plt.ylabel("Sales per month")
plt.title("Monthly Sales Data Report")
plt.legend(loc = 'upper left', fontsize = 12)
plt.grid(color = "blue", linestyle = "-.", linewidth = 1)  # This is for the background lines behind the plot or the graph
plt.xlim(1, 4)
plt.ylim(0, 2000)
plt.xticks([1, 2, 3, 4], ["M1", "M2", "M3", "M4"]) # It is used to change the indexing of the x axis
plt.show()
```

### Plotting the bar chart and label it also set the title

```python
#plt.bar(x, height, color="color name", width = value, label = "label name")

product1 = ["A", "B", "C", "D"]
sales = [1000,  1500, 800, 1200]

plt.bar(product1, sales, color='Green', label= 'Sales 2025')
plt.xlabel("Product")
plt.ylabel("Sales")
plt.title("Product Sale Companies")
plt.legend() # It shows the label'value on the left uppper corner or any location given
plt.show()
```

### Plotting the pie chart

```python
#plt.pie(values, labels=label_list, colors=color_list, autopct='%1.1f%%')

regions = ['North', 'South', 'East', 'West']
revenue = [3000, 2000, 1500,1000]

plt.pie(revenue, labels= regions, autopct='%1.1f%%', colors=['gold', 'skyblue', 'lightgreen', 'coral'])
plt.title('Revenue Contribution by Region')
plt.show()
```

### Plotting the histograms

```python
#plt.hist[data, bins=number_of_bins, colors='colorname', edgecolor='black']

scores = [23, 24, 54, 67, 84, 63, 56, 72, 20, 26, 12, 16, 37, 49, 50, 70]

plt.hist(scores, bins=5, color='purple', edgecolor='black')
plt.xlabel("Score Range")
plt.ylabel("Number of students")
plt.title("Score distirbution of Students")
# plt.grid(True)
plt.show()

```

### Understanding the use of Scatter

```python
#plt.scatter(x, y, color='colorname', marker='marker_style', label='label name')

hours_studied = [1, 2, 3, 4, 5, 6, 7, 8]
marks_score = [34,40, 45, 50, 55, 65, 80, 85]

#Scatter plot
plt.scatter(hours_studied, marks_score, color='green', marker='o', label='Student Data')
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.title("Relationship between study time and exam score")
plt.legend()
plt.grid(True) # It will show the background lining 
plt.show()
```

### Plotting the graph using the scatter plot

```python
#Scatter plot
plt.scatter([1, 2, 3], [50, 55, 60], color='blue', label='Class A')
plt.scatter([1, 2, 3], [40, 45, 50], color='orange', label='Class B')

plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.title("Comparison of the Classes")
plt.legend()
plt.grid(True)
plt.show()
```

### Plot two graph using subplot 

```python
#plt.subplot(nrows, ncols, index)

x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.subplot(1, 2, 1) #1 row, 2 column, 1st subplot
plt.plot(x,y)
plt.title("Line chart")

plt.subplot(1, 2, 2) # 1row, 2column, 2nd subplot
plt.bar(x,y)
plt.title("Bar Chart")

# plt.tight_layout()
plt.show()
```

### Using the subplots with fig, ax which gives the fig.method

```python
#fig, ax = plt.subplots(nrows, figsize=[width, height])
fig, ax = plt.subplots(1, 2, figsize=(10, 5))

x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

ax[0].plot(x,y, color='red')
ax[0].set_title("Line Plot")

ax[1].bar(x,y, color='green')
ax[1].set_title("Bar chart")

fig.suptitle("Comparison of line plot and bar chart")

plt.tight_layout()
plt.show()
```
