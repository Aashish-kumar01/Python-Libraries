```python
import numpy as np
import pandas as pd
```

```python
# Dataframe:

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eve", "Alice"],
    "Age": [25, 30, 35, np.nan, 29, 25], 
    "Department": ["HR", "IT", "Finance", "IT", "HR", "HR"],
    "Salary": [50000, 60000, 700000, 62000, np.nan, 50000]
}
```

```python
# Renaming columns:

df.rename(columns = {"Department": "Dept"}, inplace= True)
```

```python
# Data cleaning

df.isnull().sum()
```

```python
df.dropna()
```

```python
df.dropna(how = "any") # Any row that have any null value
```

```python
df.dropna(how = "all") # If alll the value in that row is null then drop it otherwise do nothing
```

```python
df["Age"].fillna(df["Age"].mean())
```

```python
df["Salary"].fillna(df["Salary"].median())
```

```python
df["Age"].fillna(method = "ffill")
```

```python
df["Age"].fillna(method = "bfill")
```

```python
df["Age"] = df["Age"].bfill()
```

```python
df
```

```python
df["Name"] = df["Name"].replace("Charlie", "Rose")
```

```python
df
```

```python
df_dup = df[df.duplicated(keep= "first")]
df_dup
```

```python
df_dup = df[df.duplicated(keep= "last")] # This will priorities last data and rather than last data if any duplicate is available this command will show that.
df_dup
```

```python
df = df.drop_duplicates()
```

```python
df
```

```python
# Invalid values:
# Lambda -> python

df["Promoted Salary"] = df["Promoted Salary"].apply(lambda x: x/10 if x > 650000 else x)
```

```python
df
```

```python
# The command given below is used to split the string any column or row but it is used when i have a string available in the dataframe that is divided by some characters like("_", " ", etc.)


# name = "alice_fernandes"
# df[["first_name", "last_name"]] = df["name"].str.split("_")
```

```python
# Apply and lambda:
def multiplying_age(x):
    return x*2

df["Age"] = df["Age"].apply(multiplying_age)
```

```python
df
```

```python
df["Age"] = df["Age"].apply(lambda x:x/2)
```

```python
df
```

```python
# Joins and merges

department_info = {
    "Dept": ["HR", "IT", "Finance"], 
    "Locatoin": ["New York", "San francisco", "Chicago"],
    "Manager": ["Laura", "Steve", "Nina"]
}

df2 = pd.DataFrame(department_info)
df2
```

```python
# Concat

pd.concat([df, df2])
```

```python
pd.concat([df, df2], axis=1)
```

```python
pd.merge(df, df2, on = "Dept")
```
